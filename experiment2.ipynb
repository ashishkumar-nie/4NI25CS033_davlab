```python
import numpy as np
import pandas as pd

sample_data = {
    "Age": [25, 30, 35, np.nan, 40, 30, 25],
    "Salary": [50000, 60000, 75000, 80000, np.nan, 60000, 50000],
    "Department": ["HR", "IT", "IT", "Finance", "HR", "IT", "HR"],
    "City": ["NY", "LA", np.nan, "Chicago", "NY", "LA", "NY"],
}

pd.DataFrame(sample_data).to_csv("data.csv", index=False)
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data.csv")

print("--- DATA LOADING & OVERVIEW ---")
print(df.head())
print(df.tail())
print(df.shape)

print("\n--- DATA TYPES & SUMMARY INFO ---")
df.info()

print("\n--- DESCRIPTIVE STATISTICS (NUMERIC) ---")
print(df.describe())

print("\n--- DESCRIPTIVE STATISTICS (CATEGORICAL) ---")
print(df.describe(include="object"))

print("\n--- INDIVIDUAL DESCRIPTIVE STATS ---")
print("Mean:\n", df.mean(numeric_only=True))
print("Median:\n", df.median(numeric_only=True))
print("Standard Deviation:\n", df.std(numeric_only=True))
print("Skewness:\n", df.skew(numeric_only=True))

print("\n--- MISSING VALUES IDENTIFICATION ---")
print("Missing mask preview:\n", df.isnull().head())
print("Total missing values per column:\n", df.isnull().sum())
print("Percentage missing values per column:\n", df.isnull().mean() * 100)

print("\n--- DUPLICATE IDENTIFICATION ---")
print("Total duplicate rows:", df.duplicated().sum())
print("Duplicate rows dataset:\n", df[df.duplicated()])
print("Duplicates based on specific subset:\n", df[df.duplicated(subset=df.columns[:2])])
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("data.csv")

df_dropped_rows = df.dropna()
df_dropped_cols = df.dropna(axis=1)

df_filled = df.copy()

num_cols = df_filled.select_dtypes(include=[np.number]).columns
df_filled[num_cols] = df_filled[num_cols].fillna(df_filled[num_cols].median())

cat_cols = df_filled.select_dtypes(include=["object", "category"]).columns
for col in cat_cols:
    df_filled[col] = df_filled[col].fillna(df_filled[col].mode()[0])

df_dedup = df.drop_duplicates()
df_dedup_subset = df.drop_duplicates(subset=df.columns[:2], keep="first")

df_cleaned = df.drop_duplicates().copy()

for col in df_cleaned.select_dtypes(include=[np.number]).columns:
    df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].median())

for col in df_cleaned.select_dtypes(include=["object", "category"]).columns:
    df_cleaned[col] = df_cleaned[col].fillna("Unknown")

df_cleaned = df_cleaned.reset_index(drop=True)

print("Original Shape:", df.shape)
print("Cleaned Shape:", df_cleaned.shape)
print("\nRemaining Missing Values:\n", df_cleaned.isnull().sum())
print("\nRemaining Duplicates:", df_cleaned.duplicated().sum())
```
