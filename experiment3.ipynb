```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

np.random.seed(42)
df = pd.DataFrame(
    {
        "Age": np.random.normal(35, 10, 250),
        "Income": np.random.normal(60000, 15000, 250),
        "Spending_Score": np.random.uniform(10, 100, 250),
        "Category": np.random.choice(["Tier 1", "Tier 2"], size=250),
    }
)
df["Savings"] = df["Income"] * 0.35 + np.random.normal(0, 3000, 250)

plt.figure(figsize=(14, 10))

plt.subplot(2, 2, 1)
sns.histplot(data=df, x="Age", kde=True, color="teal")
plt.title("Histogram: Age Distribution")

plt.subplot(2, 2, 2)
sns.boxplot(data=df, x="Category", y="Income", palette="Set2")
plt.title("Box Plot: Income by Category")

plt.subplot(2, 2, 3)
sns.scatterplot(
    data=df, x="Income", y="Savings", hue="Category", alpha=0.8
)
plt.title("Scatter Plot: Income vs. Savings")

plt.subplot(2, 2, 4)
numeric_df = df.select_dtypes(include=[np.number])
corr_matrix = numeric_df.corr()
sns.heatmap(corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()

sns.pairplot(df, hue="Category", diag_kind="kde", palette="Set1")
plt.show()
```
